Create function to create new column
New column will split the current value at the ' ' (space), then check if the letters match 'posteam' value for the row. If they do, then the yardline will be 100-(number value). If they don't, then the value will be (number value)

```python
test = data[0:99]
```

```python
# The 50 isn't owned by either team, so we need to have a condition for when drive_start_yard_line is 50.
for row in test.itertuples():
    print(f"Row drive_start_yard_line: {row.drive_start_yard_line}")
    row_off_team = row.posteam
    print(f"Row offense team: {row_off_team}")
    print(f"Row Defense team: {row.defteam}")
    try:

        if row.drive_start_yard_line == "50":
            print("Clean Yard Line: 50")
            continue

        row_yard_line_number = row.drive_start_yard_line.split(" ")[1]
        row_yard_line_team = row.drive_start_yard_line.split(" ")[0]
        if row_yard_line_team == row_off_team:
            print("Clean Yard Line: ", 100 - int(row_yard_line_number))
        else:
            print("Clean Yard Line: ", int(row_yard_line_number))
    except:
        print(f"ERROR. Row drive_start_yard_line: {row.drive_start_yard_line}",
              f"Row offense team: {row_off_team}",
              f"Row Defense team: {row.defteam}")

    print("------")
```

```python
## Turn previous cell logic into a function to create a column for clean yard line
def clean_yard_line(row):
    row_off_team = row.posteam
    try:

        if row.drive_start_yard_line == "50":
            return 50

        row_yard_line_number = row.drive_start_yard_line.split(" ")[1]
        row_yard_line_team = row.drive_start_yard_line.split(" ")[0]
        if row_yard_line_team == row_off_team:
            return 100 - int(row_yard_line_number)
        else:
            return int(row_yard_line_number)
    except:
        return None
```

```python
## Apply function to create new column
test['clean_yard_line'] = test.apply(clean_yard_line, axis=1)
test.head()
```

```python
## Test on a larger subset of data
test_2 = data[0:1000]
test_2['clean_yard_line'] = test_2.apply(clean_yard_line, axis=1)
test_2.head()
```

```python
## Function works!
data['clean_yard_line'] = data.apply(clean_yard_line, axis=1)
data.head()
```
